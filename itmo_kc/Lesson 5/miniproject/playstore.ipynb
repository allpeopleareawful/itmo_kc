```python
mapping = dict(zip(unique_playstore.columns, unique_playstore.columns.str.lower().str.replace(' ', '_')))
unique_playstore = unique_playstore.rename(columns=mapping)
unique_playstore['price'] = unique_playstore['price'].str.replace('$', '').astype(float)

unique_playstore.pivot_table(
    values = ['price', 'rating', 'reviews'],
    index = ['category', 'type'],
    aggfunc = 'mean',
)
```

```python

unique_playstore.groupby(['category', 'type']).agg(
    mean_price=('price', 'mean'),
    mean_rating=('rating', 'mean'),
    mean_reviews=('reviews', 'mean')
).round(2).to_csv('pivot.csv')
unique_playstore.groupby(['category', 'type']).agg(
    mean_price=('price', 'mean'),
    mean_rating=('rating', 'mean'),
    mean_reviews=('reviews', 'mean')).round(2)
```

```python
res = unique_playstore.pivot_table(
    values=['price', 'rating', 'reviews'],
    index=['category', 'type'],
    aggfunc='mean'
).add_prefix('mean_').round({
    'mean_price': 2,    # Цену округляем до копеек (2 знака)
    'mean_rating': 1,   # Рейтинг округляем до десятых (1 знак)
    'mean_reviews': 2   # Количество отзывов делаем целым числом (0 знаков)
}).to_csv('pivot.csv')

res
```
