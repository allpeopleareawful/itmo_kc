```python
vec1 = list(range(int(1e5)))
vec2 = list(range(int(1e5)))
```

```python
def dot_product(vec1, vec2):
    result = 0
    for i in range(len(vec1)):
        result += vec1[i] * vec2[i]
    return "{:.2e}".format(result)
```

```python
%%timeit
print(dot_product(vec1, vec2))
```

```python
import numpy as np
```

```python
vec1_np = np.array(vec1, dtype=np.float64)
vec2_np = np.array(vec2, dtype=np.float64)
```

```python
%%timeit
print(np.dot(vec1_np, vec2_np))
```
