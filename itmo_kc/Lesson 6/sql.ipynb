```python
df = pd.read_sql(
'''
    SELECT
        exp_group, 
        os, 
        COUNT("user") AS total_users, 
        MAX(age) AS max_age, 
        MIN(age) AS min_age
    FROM "user"
    GROUP BY
        exp_group, os
''',
    con=con_url
)

df.to_csv('groupby1.csv', sep=',', index=False)
```

```python
df = pd.read_sql(
'''
    SELECT f.post_id, f.time, u.age, u.os
    FROM feed_action f
    JOIN "user" u on f.user_id = u.id
    WHERE f.action = 'like' AND u.city = 'Omsk'
    ORDER BY f.time DESC
    LIMIT 100
''',    
    con=con_url
).to_csv('join1.csv', sep=',', index=False)
df
```

```python
df = pd.read_sql(
'''
    SELECT u.city, COUNT (f.action) AS views 
    FROM "user" u
    JOIN feed_action f on f.user_id = u.id
    JOIN post p on f.post_id = p.id
    WHERE u.age = 36
        AND f.action = 'view'
        AND p.topic = 'covid'
        AND f.time::date = '2021-12-01'
    GROUP BY u.city
    ORDER BY views
''',    
    con=con_url
)

df
```

```python
df = pd.read_sql(
'''
    SELECT 
        post_id,
        COUNT(*) AS likes, 
        COUNT(DISTINCT user_id) AS unique_liked_users, 
        MAX(time) AS last_like_time
    FROM feed_action
    WHERE action = 'like'
    GROUP BY post_id
    ORDER BY likes DESC
    LIMIT 20
''',    
    con=con_url
)

df
```
